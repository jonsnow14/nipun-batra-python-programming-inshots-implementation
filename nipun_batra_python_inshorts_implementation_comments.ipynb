```python
#what does ys prints ?
ys = []
for x in range(1,11):
  ys.append(x*x)
xs = [x*x for x in range (1,11)]
print(xs  == ys)
print(xs)
```

```python
#python -MAP
#map calls a function on each item
xs = ["1", "245", "129"]
print(list(map(int, xs)))
def succ(n):
  return n +1
ys  = [1,2,3]
print(list(map(succ, ys)))
```
